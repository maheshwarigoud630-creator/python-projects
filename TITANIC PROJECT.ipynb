### DATA CLEANING

### checking for null values

```python
df.isnull().sum()
```

```python
df["Age"].skew() # 62% normally distributed so replaced by mean
```

```python
df["Age"]=df["Age"].fillna(df["Age"].mean())
```

```python
df.isnull().sum()
```

### as in cabin coluum there are more than 90% of null we drop cabin colum

```python




df=df.drop("Cabin",axis=1) 
```

```python
df.isnull().sum()
```

```python
df["Embarked"]=df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df.isnull().sum()
```

### checking for outliers

```python
import numpy as np
b=df.select_dtypes(include=np.number)  ### have taken np.number as it takes both int and float values
```

```python
for i in b:
    q1=b[i].quantile(0.25)
    q2=b[i].quantile(0.50)
    q3=b[i].quantile(0.75)
    IQR=q3-q1
    uw=q3+1.5*IQR
    lw=q1-1.5*IQR
    outliers=((b[i]>uw)|(b[i]<lw))
    print(f"{i}:-",outliers.sum())
    b.loc[outliers,i]=q2
    print(f"after removing outliers:-",((b[i]>uw)|(b[i]<lw)).sum())
```

### MOST OF THEM DIED IN THE INCIDENT

 ### binning- The binning process refers to sorting items, data points, or components into distinct categories or "bins" based on specific criteria or quality tiers.

```python
bins=[0,18,40,70]
labels=["child","adult","old"]
df["AgeGroup"]=pd.cut(df["Age"],bins=bins,labels=labels)
```

```python
b=df.groupby("AgeGroup")["Survived"].mean()*100
b                                                 #pie chart
```
